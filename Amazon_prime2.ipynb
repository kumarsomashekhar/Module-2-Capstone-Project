# Project Summary

## 📊 Amazon Prime Video EDA

This project focuses on performing Exploratory Data Analysis (EDA) on the Amazon Prime Video Movies and TV Shows dataset to uncover valuable insights related to content diversity, audience targeting, regional availability, ratings, popularity, and growth trends. The dataset contains information about more than 9,000 titles available on Amazon Prime Video, along with additional details about actors, directors, and characters obtained from a separate credits dataset. By combining both datasets, the analysis provides a deeper understanding of Amazon Prime's content strategy and viewer engagement patterns.

## 🔧 Data Preparation

The project began with importing the required Python libraries such as **Pandas, NumPy, Matplotlib, and Seaborn**. The datasets `titles.csv` and `credits.csv` were loaded into Google Colab and merged using the common `id` column to create a consolidated dataset named `prime_merged`. Initial exploration was performed using functions such as `head()`, `info()`, `shape()`, and `describe()` to understand the structure, size, data types, and statistical properties of the data.

## 🧹 Data Cleaning

Data cleaning was an important part of the project. Missing values in categorical columns such as `age_certification`, `genres`, `production_countries`, `character`, and `imdb_id` were handled using placeholders like **"Unknown"**. Numeric columns such as `imdb_score`, `runtime`, `tmdb_score`, and `tmdb_popularity` were filled using median values to maintain consistency. Duplicate rows were identified and removed wherever necessary to improve the quality and reliability of the analysis.

## 📈 Key Findings

The Exploratory Data Analysis section focused on understanding Amazon Prime's content distribution and trends:

- **Movies dominate** the platform compared to TV shows, indicating Amazon Prime invests heavily in movie content while maintaining a smaller but growing collection of TV series.
- **Drama and Comedy** are the most common genres, suggesting viewers strongly prefer entertainment-focused and emotionally engaging content.
- **Rapid growth** after the year 2000 reflects the expansion of the streaming industry and Amazon Prime's aggressive content strategy.

## ⭐ Ratings Analysis

Several visualizations were created to better understand the dataset:

- **Histograms** showed that most titles are rated between 6 and 8 on IMDb
- **Count plots and pie charts** compared Movies vs Shows and analyzed age certifications
- **Scatter plots** revealed that highly popular titles are not always the highest-rated ones
- **Heatmaps** studied correlations among numerical features
- **Pair plots** provided multi-variable relationship analysis
- **Box plots** helped identify runtime variability and outliers

- The most frequently appearing actors and directors on Amazon Prime Video

- Commonly recurring character names, indicating franchise-based or recurring content

## Data Cleaning and Exploratory Data Analysis (EDA)

````python
# CHECK MISSING VALUES
df.isnull().sum()
````

````python
# Remove Duplicates
df.drop_duplicates(inplace=True)
````

##  Handling duplicate values

````python
categorical_cols = [
    'title',
    'description',
    'age_certification',
    'genres',
    'production_countries',
    'name',
    'role',
    'character',
    'imdb_id'
]

for col in categorical_cols:
    df[col].fillna(
        'Unknown',
        inplace=True
    )
````

````python
numeric_cols = [
    'runtime',
    'imdb_score',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score',
    'seasons'
]

for col in numeric_cols:
    df[col].fillna(
        df[col].median(),
        inplace=True
    )
````

## Data Visualization

Chart 1:Most Frequent Actors

````python
# Most Frequent Actors
# Find actors appearing most frequently on Amazon Prime.

top_actors = (
    df[df['role'] == 'ACTOR']['name']
    .value_counts()
    .head(10)
)

top_actors

plt.figure(figsize=(10,5))

top_actors.plot(kind='bar', color='#FF6B6B', edgecolor='black')

plt.title('Top 10 Most Frequent Actors')

plt.xlabel('Actor')

plt.ylabel('Number of Titles')

plt.xticks(rotation=45)

plt.show()

# Insights
# Some actors appear repeatedly across many titles.
# Indicates strong collaborations with Amazon Prime.

# Business Value
# identify popular actors
# understand audience attraction
# support casting decisions
````

chart 2:Most Frequent Directors

````python
# Find directors contributing most content.
top_directors = (
    df[df['role'] == 'DIRECTOR']['name']
    .value_counts()
    .head(10)
)

top_directors

plt.figure(figsize=(10,5))

top_directors.plot(kind='bar')

plt.title('Top 10 Most Frequent Directors')

plt.xlabel('Director')

plt.ylabel('Number of Titles')

plt.xticks(rotation=45)

plt.show()

# Insights
# Certain directors dominate platform content creation.

# Business Value
# Amazon can:
# maintain successful partnerships
# invest in proven creators
````

Chart 3:Most Common Characters:- fictional characters

````python
exclude = [
    'Himself',
    'Herself',
    'Self',
    'Narrator',
    'Doctor',
    'Nurse',
    'Bartender',
    'Dancer',
    'Henchman',
    'Self (archive footage)'
]

character_df = df[
    (~df['character'].isin(exclude)) &
    (df['character'].notna()) &
    (df['character'] != 'Unknown')
]

top_characters = (
    character_df['character']
    .value_counts()
    .head(10)
)

top_characters

# Visualization — Horizontal Bar Plot

plt.figure(figsize=(10,6))

top_characters.plot(kind='barh', color='#FF6B6B', edgecolor='black')

plt.title('Top 10 Most Common Characters')

plt.xlabel('Count')

plt.ylabel('Character')

plt.show()
````

Chart 3:Most Common Characters:- generic role descriptions.

````python
# Find the most recurring character names.
# Remove Missing/Unknown Characters

character_df = df[
    (df['character'].notnull()) &
    (df['character'] != 'Unknown')
]

top_characters = (
    character_df['character']
    .value_counts()
    .head(10)
)

top_characters

# Visualization — Horizontal Bar Plot

plt.figure(figsize=(10,6))

top_characters.plot(kind='barh', color='#FF6B6B', edgecolor='black')

plt.title('Top 10 Most Common Characters')

plt.xlabel('Count')

plt.ylabel('Character')

plt.show()

# Insights
# Some character names appear repeatedly across titles.
# Recurring franchises may exist.

# Business Value
# identify iconic characters
# analyze franchise popularity
````

Chart 4:Visualization — Age Certification Distribution

````python
# We analyze target audience segmentation using age certifications.
# Count plots are effective for:
# categorical audience analysis

plt.figure(figsize=(10,5))

sns.countplot(
    x='age_certification',
    data=df,
    order=df['age_certification'].value_counts().index
)

plt.title('Age Certification Distribution')

plt.xlabel('Age Certification')

plt.ylabel('Count')

plt.xticks(rotation=45)

plt.show()

# Key Insights

# The "Unknown" age certification category has the highest count, indicating that a significant portion of the dataset does not contain age certification information.
# Among the titles with known certifications, R is the most common rating, followed by PG-13 and PG, suggesting that Amazon Prime has a strong collection of content targeted toward mature and teenage audiences.
# Family-oriented certifications such as G are present but in much smaller numbers.
# TV-specific certifications (TV-MA, TV-14, TV-PG, TV-Y7, TV-Y, TV-G) represent only a small portion of the catalog compared to movie certifications.
# The presence of multiple certification categories indicates that Amazon Prime offers content suitable for viewers across different age groups.
````

````python
known_certifications = df[df['age_certification'] != 'Unknown']

plt.figure(figsize=(10,5))

sns.countplot(
    x='age_certification',
    data=known_certifications,
    order=known_certifications['age_certification'].value_counts().index
)

plt.title('Age Certification Distribution (Known Ratings Only)')
plt.xlabel('Age Certification')
plt.ylabel('Count')
plt.xticks(rotation=45)

plt.show()

# Key Insights
# 1. R-rated content dominates the catalog,The platform primarily caters to adult viewers through dramas, thrillers, crime, horror, and action movies.
# 2. PG-13 and PG are the next most common certifications.Maintaining a healthy mix of PG and PG-13 content helps attract a broader audience, including families and young adults.
# 3. Family and children's content is comparatively limited. Certifications such as G, TV-Y, TV-G, and TV-Y7 have much smaller counts.
# These ratings represent children's programming and family-friendly entertainment.
# 4. TV-specific ratings are less frequent than movie ratings. Ratings like TV-MA, TV-14, and TV-PG appear less often than movie ratings such as R, PG-13, and PG.
# 5. Diverse certifications indicate broad audience coverage
````

Chart 5:Histogram — IMDb Score Distribution

````python
# Analyze:
# rating distribution
# frequency
# spread of IMDb scores

plt.figure(figsize=(8,5))

# Plot histogram with KDE
sns.histplot(df['imdb_score'].dropna(),
             bins=20,
             kde=True,
             edgecolor='black',
             color='skyblue',
             alpha=0.7)

plt.title('IMDb Score Distribution', fontsize=14, fontweight='bold')
plt.xlabel('IMDb Score', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.show()

# Insights
# Most titles cluster between 6.0 and 8.0, indicating that the platform's content is generally well-received.
# The peak frequency likely lies around 6.5–7.0, which is typical for large movie/TV databases (average rating is often ~6.5).
# he spread (range = 2 to 10) is wide, but most variability is compressed between 5 and 8.
# The distribution appears negatively skewed (left-skewed) if the tail extends toward lower scores (2–4) and the peak is on the right.
````

Chart 6:Line Plot — Content Trend by Year

````python
# Analyze content growth over time.
# We study how Amazon Prime’s content library evolved over time.

# Goal:
# identify growth trends
# analyze expansion periods
````

````python
df['release_year'].value_counts().sort_index().tail(10)
````

````python
yearly_content = (
    df.drop_duplicates(subset='title')
    .groupby('release_year')
    .size()
)

plt.figure(figsize=(12,5))

yearly_content.plot()

plt.title('Amazon Prime Content Trend Over Years')

plt.xlabel('Release Year')

plt.ylabel('Number of Titles')

plt.show()

# Insights
# Massive growth after 2000.
# Streaming expansion accelerated content additions
# Content production increased steadily after the year 2000, with particularly rapid growth during the 2010s.
# The highest concentration of titles appears between 2018 and 2021, reflecting the streaming industry's expansion.
# The sharp decline in 2022 is most likely due to the dataset containing incomplete
# data for the latest year rather than a genuine decline in content availability.

# The dataset ends in 2022 This means dataset only contains titles up to 2022. Since 2022 was
# likely collected before the year ended, it naturally has fewer titles than previous complete years
````

Chart 7:Scatter Plot — IMDb Score vs TMDB Popularity

````python
# Analyze relationship between:
# ratings
# popularity

plt.figure(figsize=(8,5))

sns.scatterplot(
    x='imdb_score',
    y='tmdb_popularity',
    data=df
)

plt.title('IMDb Score vs TMDB Popularity')

plt.xlabel('IMDb Score')

plt.ylabel('TMDB Popularity')

plt.show()

# Insights
# Highly rated titles are not always the most popular.
# Likewise, some highly popular titles have only average IMDb ratings.
# A small number of titles act as outliers, with exceptionally high popularity despite having only moderate IMDb scores.
````

````python
correlation = df['imdb_score'].corr(df['tmdb_popularity'])

print("Correlation:", round(correlation, 2))

# The correlation coefficient is:

# Correlation = 0.07

# A value of 0.07 is very close to 0, which indicates almost no linear relationship between IMDb Score and TMDB Popularity.
````

Chart 8:Pie Chart — Movies vs Shows

````python
# Analyze platform content composition.
# Pie charts clearly show:
# proportions
# category shares

content_counts = (
    df.drop_duplicates(subset='title')['type']
    .value_counts()
)

plt.figure(figsize=(7,7))

content_counts.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Movies vs Shows')

plt.ylabel('')

plt.show()

# Insights
# Movies dominate Amazon Prime’s library.
# Amazon can diversify show content
# improve long-term subscriber engagement
````

Chart 9:Heatmap — Correlation Analysis

````python
# Analyze relationships between numeric variables.
# Heatmaps effectively display:
# correlations
# variable relationships

numeric_data = df.select_dtypes(include='number').drop(columns=['person_id'])


plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()

# Insights
# IMDb Score and TMDB Score have the strongest positive correlation (0.60).
# IMDb Votes and IMDb Score have a weak positive correlation (0.26)
# IMDb Votes and TMDB Popularity show a weak positive relationship (0.23)
# Runtime has very little influence on ratings
# Runtime vs IMDb Score: 0.10
# Runtime vs TMDB Score: 0.07.

# Amazon can:
# focus on quality rather than duration
# improve content recommendation logic
````

Chart 10:Pair Plot — Multi-variable Analysis

````python
# Pair plots help visualize:
# distributions
# trends
# relationships
# clusters

sns.pairplot(
    df[['imdb_score', 'tmdb_score', 'runtime', 'tmdb_popularity']],
    plot_kws={'alpha': 0.6, 's': 30, 'edgecolor': 'black'},
    diag_kws={'edgecolor': 'black', 'alpha': 0.7},
    palette='viridis'  # Very vibrant: 'viridis', 'plasma', 'inferno'
)

plt.suptitle('Pairplot of Numerical Features', y=1.02, fontsize=14, fontweight='bold')
plt.show()

# Insights
# IMDb Score and TMDB Score show a moderate positive relationship.
# Runtime has little relationship with ratings
# Popularity distribution is highly skewed.
# Amazon can
# identify quality-performance relationships
# improve analytics-driven content strategy
````

Chart 11:Box Plot for Runtime

````python
# Box plots are useful for:
# outlier detection
# spread analysis
# quartile visualization

plt.figure(figsize=(8,5))

sns.boxplot(
    x='type',
    y='runtime',
    data=df
)

plt.title('Runtime Distribution by Content Type')

plt.show()

# Insights
# Movies have a higher median runtime (around 95–100 minutes) compared to TV Shows, whose episode runtimes are generally 30–45 minutes.
# Movie runtimes exhibit greater variability, with several extreme outliers exceeding 200 minutes, and one title extending beyond 500 minutes.
# TV Show episode runtimes are more consistent, with most episodes concentrated between 20 and 50 minutes, indicating standardized episode lengths.
# The presence of numerous movie outliers suggests that the catalog includes extended editions, documentaries, special features, or anthology films with unusually long durations.
# The compact spread of TV Show runtimes reflects industry practices where episode lengths are typically fixed within a series.
````
